```python
import pandas as pd

# Creating a dummy dataset based on the hospital scenario
data = {
    "Patient ID": [101, 102, 103, 104, 105],
    "Name": ["John", "Sara", "Mike", "Emma", "David"],
    "Age": [45, 32, 50, 28, 61],
    "Gender": ["Male", "Female", "Male", "Female", "Male"],
    "Height": [175, 160, 180, 165, 170],
    "Weight": [70, 55, 85, 60, 78],
    "Blood Pressure": [120, 110, 140, 115, 130],
    "Glucose Level": [95, 85, 150, 90, 160],
    "Blood Group": ["A+", "B+", "O+", "AB+", "A-"],
    "Diabetes": ["No", "No", "Yes", "No", "Yes"],
}

df = pd.DataFrame(data)
print("Initial Dataset:")
print(df)
```

```python
print("\nData Types:")
print(df.dtypes)
```

```python
num_cols = df.select_dtypes(include=["number"]).columns.tolist()
cat_cols = df.select_dtypes(exclude=["number"]).columns.tolist()
```

```python
print(f"\nNumerical Columns: {num_cols}")
print(f"Categorical Columns: {cat_cols}")
```

```python
# Encoding Gender and Diabetes
df["Gender"] = df["Gender"].map({"Male": 1, "Female": 0})
df["Diabetes"] = df["Diabetes"].map({"Yes": 1, "No": 0})

# One-Hot Encoding for Blood Group
df = pd.get_dummies(df, columns=["Blood Group"], drop_first=True)

print("\nDataset after Encoding:")
print(df)
```

```python
df_cleaned = df.drop(columns=["Patient ID", "Name"])

print("\nDataset after removing Patient ID and Name:")
print(df_cleaned.head())
```

```python
print("\nSummary Statistics:")
print(df_cleaned.describe())
```

```python
df_cleaned.to_csv("cleaned_diabetes_data.csv", index=False)
print("\nCleaned dataset saved successfully as 'cleaned_diabetes_data.csv'.")
```
