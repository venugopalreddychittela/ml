```python
import numpy as np
import pandas as pd

# Sample dataset representing retail customer information with missing values
data = {
    "CustomerID": [101, 102, 103, 104, 105, 106],
    "Age": [25, np.nan, 35, 40, np.nan, 28],
    "Income": [50000, 60000, np.nan, 80000, 55000, np.nan],
    "City": ["New York", "London", np.nan, "London", "London", "Paris"],
    "Membership": ["Gold", "Silver", "Gold", np.nan, "Silver", "Silver"],
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)
```

```python
# Boolean mask showing True where values are missing
print("\nMissing Values Mask:")
print(df.isnull())
```

```python
print("\nMissing Values Count per Column:")
print(df.isnull().sum())
```

```python
# Create a copy to track changes before and after
df_cleaned = df.copy()

# Replace missing values with Mean for numerical columns
df_cleaned["Age"].fillna(df_cleaned["Age"].mean(), inplace=True)
df_cleaned["Income"].fillna(df_cleaned["Income"].mean(), inplace=True)

```

```python
df_cleaned.fillna({"Age":df_cleaned["Age"].mean()}, inplace=True)
df_cleaned.fillna({"Income":df_cleaned["Income"].mean()}, inplace=True)
print(df_cleaned)
```

```python
print("--- BEFORE PREPROCESSING ---")
print(df)
print("\nMissing count before:")
print(df.isnull().sum())

print("\n" + "=" * 40 + "\n")

print("--- AFTER PREPROCESSING ---")
print(df_cleaned)
print("\nMissing count after:")
print(df_cleaned.isnull().sum())
```
